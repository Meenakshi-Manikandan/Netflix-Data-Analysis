```python
def genre_analysis(df):
    genres = df['listed_in'].str.split(', ').explode()
    top_genres = genres.value_counts().head(10)
    return top_genres
```

```python
genre_analysis(df)
```

```python
def country_analysis(df):
    countries = df['country'].str.split(', ').explode()
    top_countries = countries[countries != 'Unknown'].value_counts().head(10)
    return top_countries
```

```python
def create_bar_chart(data, title, xlabel, ylabel):
    data.plot(kind='bar', figsize=(10,5))

    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.xticks(rotation=45)

    plt.show()
```

```python
create_bar_chart(
    top_10_genres,
    'Top 10 Netflix Genres',
    'Genres',
    'Number of Titles'
)
```

```python
def dashboard_report(df):

    genres = df['listed_in'].str.split(', ').explode()
    top_genres = genres.value_counts().head(10)

    countries = df['country'].str.split(', ').explode()
    top_countries = countries[countries != 'Unknown'].value_counts().head(10)

    ratings = df['rating'].value_counts().head(10)

    content = df['type'].value_counts()

    plt.figure(figsize=(14,10))

    plt.subplot(2,2,1)
    top_genres.plot(kind='bar')
    plt.title('Top 10 Genres')
    plt.xticks(rotation=45)

    plt.subplot(2,2,2)
    top_countries.plot(kind='bar')
    plt.title('Top 10 Countries')
    plt.xticks(rotation=45)

    plt.subplot(2,2,3)
    ratings.plot(kind='bar')
    plt.title('Content Ratings')
    plt.xticks(rotation=45)

    plt.subplot(2,2,4)
    content.plot(kind='pie', autopct='%1.1f%%')
    plt.title('Movies vs TV Shows')
    plt.ylabel('')

    plt.tight_layout()
    plt.show()
```

```python
dashboard_report(df)
```

## 1. Problem Statement

The objective of this project is to analyze Netflix content using Python. The analysis focuses on understanding content distribution, popular genres, country-wise production, content ratings, and yearly and monthly trends.

The project uses Python libraries such as Pandas, NumPy, Matplotlib, and Seaborn to clean, analyze, and visualize the Netflix dataset.

The main goal is to generate meaningful business insights and recommendations based on the analysis.

## 2. Dataset Overview

The Netflix dataset contains information about movies and TV shows available on Netflix.

The dataset includes the following columns:

- show_id – Unique ID for each title.
- type – Movie or TV Show.
- title – Name of the content.
- director – Director of the content.
- cast – Actors involved.
- country – Country of production.
- date_added – Date when content was added to Netflix.
- release_year – Year of release.
- rating – Content rating.
- duration – Movie duration or TV show seasons.
- listed_in – Genre or category.
- description – Short description of the content.

**Dataset Size:**

The dataset was used to perform data cleaning, exploratory data analysis, data visualization, and business analysis.

## 3. Analysis

The Netflix dataset was analyzed using Python.

The following analyses were performed:

1. **Genre-wise Analysis:** Identified the most common genres available on Netflix.
2. **Country-wise Analysis:** Identified the countries contributing the most content.
3. **Rating-wise Analysis:** Analyzed the distribution of content ratings.
4. **Movie vs TV Show Analysis:** Compared the number of movies and TV shows.
5. **Year-wise Analysis:** Studied content distribution based on release years.
6. **Date Analysis:** Analyzed yearly and monthly content additions.

These analyses helped identify patterns and trends in Netflix's content library.

## 5. Business Insights

Based on the Netflix data analysis, the following business insights were identified:

1. **Popular Genres:** International Movies, Dramas, and Comedies are among the leading genres in the dataset.

2. **Top Producing Country:** The United States contributes the highest number of titles.

3. **Content Ratings:** TV-MA is the most common content rating.

4. **Content Type:** Movies dominate the Netflix content library compared to TV Shows.

5. **Content Distribution:** Movies account for approximately 69.6% of the content, while TV Shows account for 30.4%.

6. **Release Year Trends:** The number of titles varies across different release years.

7. **Content Additions:** Netflix content additions show changes across different years.

8. **Monthly Trends:** Content additions vary across different months.

9. **Movie Duration:** Movie durations differ, with most movies having durations around 90–120 minutes.

10. **Content Diversity:** Netflix offers content from different countries, genres, and rating categories.

These findings help understand the distribution and characteristics of Netflix's content library.

## 7. Business Recommendations

Based on the analysis, the following recommendations can be made:

1. **Focus on Popular Genres:** Netflix can continue developing content in popular genres to maintain a diverse content library.

2. **Expand Country-wise Content:** Netflix can include more content from different countries to improve its international content collection.

3. **Improve Content Diversity:** Netflix can maintain a balance between movies and TV shows to provide different entertainment options.

4. **Focus on Content Ratings:** Netflix can offer content suitable for different audience groups based on content ratings.

5. **Plan Content Releases:** Netflix can study yearly and monthly content addition trends to plan future content releases more effectively.

These recommendations are based on the available Netflix dataset. Actual viewer preferences and demand would require additional audience data.
